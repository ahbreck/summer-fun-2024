##Assessing Class Imbalance/Balance
Imbalance could result in poor performance in a neural network for the class(es) with low representation, which will also degrade overall performance. Source: https://medium.com/geekculture/eda-for-image-classification-dcada9f2567a

```python
# create a function to summarize distribution by species or some other grouping

# helper function to make pivot table
def make_summary(df, grouping):
    # create a pivot table to summarize the number of images in the training dataset for each group
    count_by_group = pd.pivot_table(df, index = grouping, aggfunc = 'count')
    count_by_group = count_by_group[['image_id']].reset_index().rename(columns = {'image_id': 'count'})
    return count_by_group

# main function
def view_by_group(df, grouping):
  # call the helper function to make summary for plotting
  count_by_group = make_summary(df, grouping)

  # plot a histogram to see how balanced the representation is across the grouping
  count_by_group['count'].plot(kind='hist', bins=20, title='Histogram of Number of Images per ' + grouping)
  plt.xlabel('Number of Images')
  plt.ylabel('Count of ' + grouping)
  plt.gca().spines[['top', 'right',]].set_visible(False)
```

As discussed above, we can augment our training data by generating mutated copies of images (e.g. randomly resized, rescaled, rotated, flipped, etc.) and adding them to the pool of images to train on. See page 315 (Chapter 11) of Hands-On Learning with Scikit-Learn & TensorFlow by Aurelien Geron. See also: https://www.geeksforgeeks.org/python-data-augmentation/. Source for the below code: https://www.kaggle.com/code/hsinwenchang/keras-mobilenet-data-augmentation-visualize

```python
nb_classes = 1010
batch_size = 128
img_size = 128
nb_epochs = 10
```

```python
%%time
# Create ImageDataGenerator object, which specifies the range within images can be modified in any given respect.
# For example, rotation_range argument here specifies that images can be rotated anywhere from 0 to 45 degrees in either direction
train_datagen=ImageDataGenerator(rescale=1./255,
                                 rotation_range=45,
                                 width_shift_range=.15,
                                 height_shift_range=.15,
                                 horizontal_flip=True,
                                 zoom_range=0.5)

# Create a generator based on the above, specifying where to access the info, and how much to access at a time.
train_generator=train_datagen.flow_from_dataframe(
    # This is the dataframe that includes the filenames for all our training images as well as their corresponding species label (target)

    # CHANGED
    # dataframe=train_labels_simple,
    dataframe=amphibians_train_labels_simple,

    # This is the directory where the images are located
    directory=input_dir,
    # This is the column in the dataframe with the image filenames
    x_col="file_name",
    # This is the column in the dataframe with the species label (target)
    y_col="category_id",
    # This describes how many images to process at a time. Above we already specified this as 128
    batch_size=batch_size,
    shuffle=True,
    class_mode="sparse",
    target_size=(img_size,img_size))
```

## Random Forest

We can use the features extracted from the ENB0 model to build our Random Forest Classifier

```python
# Extract features
train_features = ENB0model.predict(train_generator_resnet50_cleaned)
validation_features = ENB0model.predict(val_generator_resnet50_cleaned)

# Get labels
train_labels = train_generator_resnet50_cleaned.classes
validation_labels = val_generator_resnet50_cleaned.classes

# Encode labels
label_encoder = LabelEncoder()
train_labels_encoded = label_encoder.fit_transform(train_labels)
validation_labels_encoded = label_encoder.transform(validation_labels)
```

```python
%%time
# Build and train Random Forest model
rfmodel = RandomForestClassifier(n_estimators=100, random_state=42)
rfmodel.fit(train_features, train_labels_encoded)

# Make predictions
train_predictions = rfmodel.predict(train_features)
validation_predictions = rfmodel.predict(validation_features)

# Calculate accuracy
train_accuracy = accuracy_score(train_labels_encoded, train_predictions)
validation_accuracy = accuracy_score(validation_labels_encoded, validation_predictions)

print(f"Train Accuracy: {train_accuracy}")
print(f"Validation Accuracy: {validation_accuracy}")
```
